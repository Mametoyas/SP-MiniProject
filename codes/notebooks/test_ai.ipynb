## จำลองเล่นเกม Hangman ทีละตัว

```python
MAX_WRONG = 6

def play(secret: str):
    secret  = secret.lower()
    guessed = set()
    wrong   = set()

    for turn in range(1, 100):
        pattern = "".join(c if c in guessed else "_" for c in secret)

        if "_" not in pattern:
            print(f"✅ WIN in {turn-1} turns! word = {secret}")
            return

        if len(wrong) >= MAX_WRONG:
            print(f"❌ LOSE — word was: {secret}")
            return

        suggest = predict(pattern, guessed, wrong)
        guess   = suggest[0]          # AI เลือกตัวแรกเสมอ

        if guess in secret:
            guessed.add(guess)
            result = "✓"
        else:
            wrong.add(guess)
            result = "✗"

        print(f"turn {turn:2d} | pattern: {pattern:15s} | guess: {guess} {result} | wrong({len(wrong)}): {sorted(wrong)}")
```

```python
play("banana")
```

## ทดสอบหลายคำพร้อมกัน

```python
test_words = ["elephant", "python", "banana", "tiger", "brazil", "strawberry"]

for word in test_words:
    play(word)
    print()
```

## Win Rate Test (random sample)

```python
import random
with open("../../datasets/words.txt", encoding="utf8") as f:
    all_words = [w.strip().lower() for w in f if w.strip().isalpha()]

sample = random.sample(all_words, min(200, len(all_words)))

MAX_WRONG = 6
wins = 0

for secret in sample:
    guessed, wrong = set(), set()
    for _ in range(100):
        pattern = "".join(c if c in guessed else "_" for c in secret)
        if "_" not in pattern:
            wins += 1
            break
        if len(wrong) >= MAX_WRONG:
            break
        guess = predict(pattern, guessed, wrong)[0]
        (guessed if guess in secret else wrong).add(guess)

print(f"Win Rate: {wins}/{len(sample)} = {wins/len(sample)*100:.2f}%")
```
